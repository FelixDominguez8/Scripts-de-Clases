```python
#Corpus Español
import nltk
nltk.download("cess_esp")
from nltk.corpus import cess_esp
sentencias = cess_esp.sents()[:300]
sentencias[:2]
```

```python
#Normalizacion basica
import re
def limpiar_tokens(tokens):
  texto = " ".join(tokens).lower()
  texto = re.sub(r"[^a-záéíóúñü\s]", "", texto)
  return texto.split()

tokens_limpios = [limpiar_tokens(s) for s in sentencias]
tokens_limpios[:2]
```

```python
#Cuenta cuantos tokens tiene cada oracion
longitudes = [len(s) for s in tokens_limpios]
longitudes[:10]
```

```python
#Obtener las palabras mas frecuentes del corpus
from collections import Counter
todas = []
for s in tokens_limpios:
  todas.extend(s)
Counter(todas).most_common(10)
```

```python
#Representacion del texto
textos = [" ".join(s) for s in tokens_limpios]
textos[:2]
```

```python
#BoW
from sklearn.feature_extraction.text import CountVectorizer
bow = CountVectorizer(max_features=2000)
X_bow = bow.fit_transform(textos)
X_bow.shape
```

```python
#Convierte la matriz BoW en DataFrame.
import pandas as pd
bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=bow.get_feature_names_out()
)
bow_df.head()
```

```python
#TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=2000)
X_tfidf = tfidf.fit_transform(textos)
```

```python
#Comparar el peso TF-IDF de la palabra "gobierno".
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)
tfidf_df["gobierno"].head()
```

```python
#Modelos NLP
#Etiquetas
y = [1 if len(s) > 10 else 0 for s in tokens_limpios]
```

```python
#Training/Testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.3, random_state=42
)
```

```python
#Entrenar Naive Bayes y calcula accuracy.
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
nb = MultinomialNB()
nb.fit(X_train, y_train)
accuracy_score(y_test, nb.predict(X_test))
```

```python
#Entrenar Logistic Regression y compara resultados.
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
accuracy_score(y_test, lr.predict(X_test))
```

```python
#Preparacion de tokens
tokens_filtrados = [s for s in tokens_limpios if len(s) > 5]
```

```python
#Buscar palabras similares a "presidente".
model.wv.most_similar("presidente")
```

```python
#Calcular similitud entre "gobierno" y "estado"
model.wv.similarity("gobierno", "estado")
```

```python
#Pipeline
from sklearn.pipeline import Pipeline
pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=3000)),
    ("clf", LogisticRegression(max_iter=1000))
])
```

```python
#Training
pipeline.fit(textos, y)
```

```python
#Predecir la clase del texto:
#"el gobierno aprobo una nueva ley"
pipeline.predict(["el gobierno aprobo una nueva ley"])
```
