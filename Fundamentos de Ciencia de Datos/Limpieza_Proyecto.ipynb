```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
df = pd.read_csv('./Datasets/TelcoClientesFuga.csv')
```

```python
df = df[df['TotalCharges'].str.strip() != '']
df = df.reset_index(drop=True)
```

```python
dummies = pd.get_dummies(df['MultipleLines'], prefix='MultipleLines').astype(int)
dummies = dummies.drop('MultipleLines_No phone service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('MultipleLines', axis=1)
```

```python
dummies = pd.get_dummies(df['OnlineSecurity'], prefix='OnlineSecurity').astype(int)
dummies = dummies.drop('OnlineSecurity_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('OnlineSecurity', axis=1)
```

```python
dummies = pd.get_dummies(df['OnlineBackup'], prefix='OnlineBackup').astype(int)
dummies = dummies.drop('OnlineBackup_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('OnlineBackup', axis=1)
```

```python
dummies = pd.get_dummies(df['DeviceProtection'], prefix='DeviceProtection').astype(int)
dummies = dummies.drop('DeviceProtection_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('DeviceProtection', axis=1)
```

```python
dummies = pd.get_dummies(df['TechSupport'], prefix='TechSupport').astype(int)
dummies = dummies.drop('TechSupport_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('TechSupport', axis=1)
```

```python
dummies = pd.get_dummies(df['StreamingTV'], prefix='StreamingTV').astype(int)
dummies = dummies.drop('StreamingTV_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('StreamingTV', axis=1)
```

```python
dummies = pd.get_dummies(df['StreamingMovies'], prefix='StreamingMovies').astype(int)
dummies = dummies.drop('StreamingMovies_No internet service', axis=1)
df = pd.concat([df, dummies], axis=1)
df = df.drop('StreamingMovies', axis=1)
```

```python
dummies = pd.get_dummies(df['InternetService'], prefix='InternetService').astype(int)
df = pd.concat([df, dummies], axis=1)
df = df.drop('InternetService', axis=1)
```

```python
df['Partner'] = df['Partner'].map({'Yes': 1, 'No': 0})
df['Dependents'] = df['Dependents'].map({'Yes': 1, 'No': 0})
df['PhoneService'] = df['PhoneService'].map({'Yes': 1, 'No': 0})
df['PaperlessBilling'] = df['PaperlessBilling'].map({'Yes': 1, 'No': 0})
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

```python
df = df.select_dtypes(include=[np.number])
df.columns = df.columns.str.strip()
```

```python
matriz_correlacion = df.corr()
```

```python
caracteristicas_alta_correlacion = [(col1, col2, matriz_correlacion.loc[col1,col2])
                                   for col1 in matriz_correlacion.columns
                                   for col2 in matriz_correlacion.columns
                                   if col1 != col2 and abs(matriz_correlacion.loc[col1,col2])>0.8]
```

```python
multicolinealidad_df = pd.DataFrame(caracteristicas_alta_correlacion, columns = ["Caracteristica 1","Caracteristica 2","Correlacion"])
print("?\nCaracteristicas altamente correlacionadas:\n", multicolinealidad_df)
```
