```python
for indice in range(len(dfLinkedinCanada.salary)):
    if type(dfLinkedinCanada.salary[indice]) is not str: 
        dfLinkedinCanada.drop(indice, inplace=True)
```

```python
dfLinkedinCanada
```

```python
dfLinkedinCanada.drop(32, inplace=True)
dfLinkedinCanada
```

```python
low_salary = []
high_salary = []
average_salary = []
for salary in dfLinkedinCanada.salary:
    split_salary = salary.split('-')
    replaced_salary1 = float(split_salary[0].replace('CA$','').replace(",",""))
    replaced_salary2 = float(split_salary[1].replace('CA$','').replace(",",""))
    average = (replaced_salary1 + replaced_salary2) / 2
    low_salary.append(replaced_salary1)
    high_salary.append(replaced_salary2)
    average_salary.append(average)
```

```python
dfLinkedinCanada['LOW_SALARY'] = low_salary
dfLinkedinCanada['HIGH_SALARY'] = high_salary
dfLinkedinCanada['AVERAGE_SALARY'] = average_salary
```
