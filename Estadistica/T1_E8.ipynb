Ingresar los datos

```r
datos <- c(6,24,91,8,4,25,3,83,89,34,25,24,18,6,23,10,28,4,63,6,5,40,2,22,23,
           6,23,26,44,49,34,2,33,9,16,55,23,13,23,4,8,26,70,4,6,60,23,95,28,
           49,6,57,33,56,7)
```

Hacer las divisiones

```r
breaks <- c(0, 15, 30, 45, 60, 75, 90, 96)
```

Mostrar el histograma

```r
hist(datos,
     breaks = breaks,
     col = "skyblue",
     main = "Histograma de Primeros Goles en el Mundial de Futbol de 2006",
     xlab = "Rango de valores",
     ylab = "Frecuencia",
     freq = TRUE,
     right = FALSE)
```

Graficar ECDF (Originales)

```r
plot(ecdf(datos),
     main = "ECDF - Datos originales",
     xlab = "Valor",
     ylab = "Frecuencia acumulada",
     verticals = TRUE,
     do.points = FALSE)
```

Ahora con los agrupados

```r
histograma <- hist(datos, breaks = breaks, plot = FALSE)

frec_relativa <- histograma$counts / sum(histograma$counts)
frec_acumulada <- cumsum(frec_relativa)

x_vals <- histograma$breaks[-1]

plot(x_vals, frec_acumulada, type = "s",
     main = "ECDF - Datos agrupados",
     xlab = "Valor (limites superiores)",
     ylab = "Frecuencia acumulada",
     ylim = c(0, 1))
```
