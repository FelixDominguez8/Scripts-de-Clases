```r
categorias <- cut(distancia,
                  breaks = c(5, 15, 20, 30),
                  labels = c("corta", "moderada", "larga"),
                  right = TRUE)
```

```r
tabla_freq <- table(categorias)
```

```r
marcas <- c(10, 17.5, 25)
f <- as.numeric(tabla_freq)
media_ponderada <- sum(marcas * f) / sum(f)
media_ponderada
```

```r
N <- sum(f)
F_acum <- cumsum(f)
clases <- c(5,15,20,30)

clase_mediana_index <- which(F_acum >= N/2)[1]

L <- clases[clase_mediana_index]
f_m <- f[clase_mediana_index]
a <- clases[clase_mediana_index + 1] - L
F_ant <- ifelse(clase_mediana_index==1, 0, F_acum[clase_mediana_index-1])

mediana_ponderada <- L + ((N/2 - F_ant)/f_m)*a
mediana_ponderada
```
