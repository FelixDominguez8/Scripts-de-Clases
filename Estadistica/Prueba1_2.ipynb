```r
library(readr)
transporte<-read.csv('/content/sample_data/encuesta_transporte_publico_completa.csv')
```

```r
attach(transporte)
hist(Edad ,xlab = 'Edad', , ylab = 'Frecuencia', xlim = c(15,80), col = 'lightblue')
lines(c(35,35), c(0,500),type = 'l', lty = 2,lwd = 3)
```

b)

```r
install.packages("ggplot2")
library(ggplot2)
```

```r
g2 <- ggplot(data = transporte, aes(x = Edad)) + geom_histogram(fill = 'darkblue', alpha = 0.5, binwidth = 2.5)+ scale_y_continuous(name = 'Edad')
plot(g2)
```
