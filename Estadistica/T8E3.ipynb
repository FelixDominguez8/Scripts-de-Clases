```r
no_subasta <- c(18.19, 16.98, 19.97, 16.98, 18.19, 15.99, 13.79,
                15.90, 15.90, 15.90, 15.90, 15.90, 19.97, 17.72)

subasta <- c(10.50, 12.00, 9.54, 10.55, 11.99, 9.30, 10.59, 10.50,
             10.01, 11.89, 11.03, 9.52, 15.49, 11.02)

alpha <- 0.05

desc <- function(x){
  media <- mean(x)
  varianza <- var(x)
  sd <- sd(x)
  cv <- sd/media
  return(list(Media=media, Varianza=varianza, DesviacionEstandar=sd, CoefVariacion=cv))
}

desc(no_subasta)
desc(subasta)

res2 <- t.test(no_subasta, mu=16.95)
res2
if(res2$p.value < alpha) "Rechazo H0" else "No rechazo H0"

cat("\n")

res3 <- t.test(no_subasta, conf.level=1-alpha)
res3$conf.int

res4 <- t.test(subasta, mu=16.95, alternative="less")
res4
if(res4$p.value < alpha) "Rechazo H0" else "No rechazo H0"

res5i <- t.test(no_subasta, subasta, alternative="greater", var.equal=TRUE)
res5i
if(res5i$p.value < alpha) "Rechazo H0" else "No rechazo H0"

res5ii <- t.test(no_subasta, subasta, alternative="greater", var.equal=FALSE)
res5ii
if(res5ii$p.value < alpha) "Rechazo H0" else "No rechazo H0"
```
