a) Represente graficamente la funcion de distribucion acumulada con R (colab)

```r
funcionCDF <- function(x) {
  ifelse(x < 0, 0,
         ifelse(x >= 0 & x <= 1, 3*x^2 - 2*x^3,
                ifelse(x > 1, 1, NA)))
}

x_values <- seq(-0.5, 1.5, by = 0.01)

y_values <- funcionCDF(x_values)

plot_data <- data.frame(x = x_values, y = y_values)

plot(plot_data$x, plot_data$y, type = "l",
     main = "Funcion de distribucion acumulativa",
     xlab = "x",
     ylab = "F(x)")
```
