```r
library(readr)
datos<-read.csv('/content/sample_data/encuesta_transporte_publico_completa.csv')
```

1)

```r
sapply(datos[, c("TiempoViaje_min", "Edad", "IngresoMensual")], mean)

sapply(datos[, c("TiempoViaje_min", "Edad", "IngresoMensual")], median)

sapply(datos[, c("TiempoViaje_min", "Edad", "IngresoMensual")], min)

sapply(datos[, c("TiempoViaje_min", "Edad", "IngresoMensual")], max)

sapply(datos[, c("TiempoViaje_min", "Edad", "IngresoMensual")], function(x) quantile(x, probs = c(0.25, 0.75)))
```

2)

```r
quantile(datos$TiempoViaje_min, 0.95)
quantile(datos$IngresoMensual, 0.95)
```

3)

```r
desviacion_media_abs <- function(x) {
  mean(abs(x - median(x)))
}

desviacion_media_abs(datos$Edad)
```

4)

```r
tiempo_viaje_scaled <- scale(datos$TiempoViaje_min)

mean(tiempo_viaje_scaled)
var(tiempo_viaje_scaled)
```

5)

```r
boxplot(datos$TiempoViaje_min, outline = FALSE, main = "Tiempo de viaje")
boxplot(datos$IngresoMensual, outline = FALSE, main = "Ingreso mensual")
```

6)

```r
intervalos <- cut(datos$TiempoViaje_min, breaks = seq(floor(min(datos$TiempoViaje_min)), ceiling(max(datos$TiempoViaje_min)), by = 10), right = FALSE)

freq <- table(intervalos)

limites <- as.numeric(sub("\\[|\\)", "", unlist(strsplit(names(freq), ","))))
puntos_medios <- (limites[seq(1, length(limites), by=2)] + limites[seq(2, length(limites), by=2)])/2

media_ponderada <- sum(puntos_medios * as.numeric(freq)) / sum(freq)
media_ponderada
```
