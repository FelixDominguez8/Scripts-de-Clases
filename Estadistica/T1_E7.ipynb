Crear Dataframe

```r
elecciones <- data.frame(
  Partido = c("ANC", "DA", "EFF", "IFP", "COPE", "Otros"),
  Resultados2014 = c(62.15, 22.23, 6.35, 2.40, 0.67, 6.20)
)
```

Hacer grafico de barras

```r
barplot(elecciones$Resultados2014,
        names.arg = elecciones$Partido,
        col = "skyblue",
        main = "Resultados por Partido Politico",
        xlab = "Partido Politico",
        ylab = "Porcentaje de Votos")
```

Crear otro dataframe pero con las elecciones de 2009

```r
elecciones2 <- data.frame(
  Partido = c("ANC", "DA", "EFF", "IFP", "COPE", "Otros"),
  Resultados2014 = c(62.15, 22.23, 6.35, 2.40, 0.67, 6.20),
  Resultados2009 = c(65.90, 16.66, 0, 4.55, 7.42, 5.47)
)
```

Crear grafica de barras doble

```r
resultados <- as.matrix(elecciones2[, c("Resultados2014", "Resultados2009")])
rownames(resultados) <- elecciones2$Partido
barplot(t(resultados),
        beside = TRUE,
        col = c("skyblue", "orange"),
        main = "Comparación de Resultados Electorales",
        xlab = "Partido Político",
        ylab = "Porcentaje de Votos",
        legend.text = c("2014", "2009"))
```
