```r
set.seed(123)

n_sample <- 1000
n_rep <- 1000

sample_means <- replicate(n_rep, mean(rnorm(n_sample, mean=0, sd=1)))

mean_of_means <- mean(sample_means)
var_of_means <- var(sample_means)

dens <- density(sample_means)
x_grid <- seq(min(dens$x), max(dens$x), length.out=400)
theo_density <- dnorm(x_grid, mean=0, sd=sqrt(1/n_sample))

plot(dens, main="Densidad kernel de medias",
     xlab="media muestral", ylab="densidad")
lines(x_grid, theo_density, lty=2, lwd=2)
legend("topright", legend=c("Kernel empirica","N(0, 1/n) teorica"),
       lty=c(1,2), bty="n")

cat("Media =", round(mean_of_means,6), "\n")
cat("Varianza =", round(var_of_means,6), "\n")
cat("Varianza teorica =", round(1/n_sample,6), "\n")
```

```r
set.seed(123)

n_sample <- 1000
n_rep <- 1000

sample_means_exp <- replicate(n_rep, mean(rexp(n_sample, rate=1)))

mean_of_means_exp <- mean(sample_means_exp)
var_of_means_exp <- var(sample_means_exp)

dens_exp <- density(sample_means_exp)
x_grid_exp <- seq(min(dens_exp$x), max(dens_exp$x), length.out=400)
theo_density_exp <- dnorm(x_grid_exp, mean=1, sd=sqrt(1/n_sample))

plot(dens_exp, main="Densidad kernel de medias",
     xlab="media muestral", ylab="densidad")
lines(x_grid_exp, theo_density_exp, lty=2, lwd=2)
legend("topright", legend=c("Kernel empirica","N(1, 1/n) teorica"),
       lty=c(1,2), bty="n")

cat("Media =", round(mean_of_means_exp,6), "\n")
cat("Varianza =", round(var_of_means_exp,6), "\n")
cat("Varianza Teorica =", round(1/n_sample,6), "\n")
```
