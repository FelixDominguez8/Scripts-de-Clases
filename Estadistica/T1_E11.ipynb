Crear Dataset

```r
datos <- c(
  1,
  rep(2, 3),
  rep(3, 8),
  rep(4, 8),
  rep(5, 27),
  rep(6, 30),
  rep(7, 11),
  rep(8, 6),
  rep(9, 4),
  rep(10, 2)
)
```

Crear histograma

```r
plot(ecdf(datos),
     main = "ECDF",
     xlab = "Respuestas",
     ylab = "Frecuencia acumulada",
     verticals = TRUE,
     do.points = FALSE)
```
