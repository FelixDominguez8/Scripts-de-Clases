```r
library(readr)
datos<-read.csv('/content/sample_data/theatre.csv')
```

1.

```r
culture_hombres <- datos$Culture[datos$Sex == 0]
culture_mujeres <- datos$Culture[datos$Sex == 1]

res1 <- t.test(culture_hombres, culture_mujeres, alternative="two.sided", var.equal = FALSE)
print(res1)
if(res1$p.value < 0.05) {
  print("Rechazamos H0 ya que hombres y mujeres gastan diferente en cultura.")
} else {
  print("No rechazamos H0 ya que no hay evidencia de diferencia en gasto cultural.")
}
```

3.

```r
theatre_hombres <- datos$Theatre[datos$Sex == 0]
theatre_mujeres <- datos$Theatre[datos$Sex == 1]

res2 <- t.test(theatre_mujeres, theatre_hombres, alternative="greater", var.equal = FALSE)
print(res2)
if(res2$p.value < 0.05) {
  print("Rechazamos H0 ya que las mujeres gastan mas en teatro que los hombres.")
} else {
  print("No rechazamos H0 ya que no hay evidencia de que las mujeres gasten mas.")
}
```

4.

```r
res3 <- t.test(datos$Theatre, datos$Theatre_ly, alternative="two.sided", paired = TRUE)
print(res3)
if(res3$p.value < 0.05) {
  print("Rechazamos H0 ya que el gasto en teatro cambio respecto al año anterior.")
} else {
  print("No rechazamos H0 ya que no hay evidencia de cambio en el gasto en teatro.")
}
```
