3.

```r
MAD_Funcion <- function(x) {
  mean(abs(x - mean(x)))
}
```

```r
Temperatura_MAD <- MAD_Funcion(tutor$temperature)
Temperatura_MAD
```

A base de la respuesta podemos entender de que el lugar en donde se realizan estas tutorias es uno con temperaturas bastante poco cambiantes solo con un margen de 4 grados desde la media

```r
tiempo_agrupado <- cut(tutor$time, breaks=seq(8, 20, by=0.5), right=FALSE)
```

```r
frecuencia <- table(tiempo_agrupado)
```

```r
separaciones <- seq(8.25, 19.75, by=0.5)
```

```r
media_ponderada <- sum(separaciones * as.numeric(frecuencia)) / sum(frecuencia)
media_ponderada
```

```r
mean(tutor$time)
```

A base de lo que podemos ver ambas la media y la media ponderada, son muy similares con una diferencia de de menos de 0.01, lo cual dice que los intervalo son bastante representivos de la data en general
