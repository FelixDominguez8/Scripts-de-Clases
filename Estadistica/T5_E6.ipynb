```r
p <- 0.125
n <- 1:200
prob <- 1 - pbinom(2, n, p)

plot(n, prob, type="l", lwd=2, col="blue",
     xlab="Numero de boletos comprados",
     ylab="Probabilidad de ≥ 3 premios")
abline(h=0.99, col="red", lty=2)
```
