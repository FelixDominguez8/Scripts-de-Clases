```r
distancia <- c(1,1,1,1,2,2,3,3)
```

```r
moda <- function(x) {
  tabla <- table(x)
  max_freq <- max(tabla)
  moda_valores <- names(tabla)[tabla == max_freq]
  if(is.numeric(x)) {
    moda_valores <- as.numeric(moda_valores)
  }
  return(moda_valores)
}
```

```r
moda(distancia)
```
